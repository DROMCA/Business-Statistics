### Question 1 (Easy)

Alpha Repairs records parts used and the elapsed completion time of each job. Each listed part was used once.

|---|---|---:|
| J01 | Battery | 30 |
| J01 | Cable | 30 |

b. Can the original register be used directly to count each type of part used? Explain.

---

### Question 1: Solution

a.

| Job | Completion time (minutes) |
|---|---:|
| J01 | 30 |
| J02 | 90 |

One observation represents one repair job. The original register gives J01 three rows and J02 one row. Averaging those rows would give the shorter job three times the weight and understate the average completion time per job.

b. Yes. Each row records one part used once, so the original register can be used directly to count parts. Battery, cable, connector, and screen each appear once.

The appropriate observation depends on the question: a job for completion times, a part used for part counts.

### Question 2 (Easy)

Beta Service Centres records these variables:

| Variable | Example value | Meaning |
|---|---|---|
| centre_code | 104 | Identifier assigned to the centre |
| service_rating | 4 | Customer chooses one category: very poor = 1, poor = 2, fair = 3, good = 4, very good = 5 |

a. Classify each variable in a table: categorical or quantitative, discrete or continuous where applicable, and measurement scale.

| Statement | Comparison |
|---|---|
| A | A centre at 20°C is 5°C warmer than one at 15°C. |
| B | A rating of 4 indicates twice the service quality of a rating of 2. |
| C | A waiting time of 25 minutes is twice a waiting time of 12.5 minutes. |

a.

### Question 3 (Average)

Gamma Retail records monthly sales at two kiosks:

| Kiosk | Month | Sales (thousand MAD) |
|---|---|---:|
| Alpha | January | 40 |
| Alpha | February | 60 |
| Alpha | March | 80 |
| Beta | January | 80 |
| Beta | February | 60 |
| Beta | March | 40 |

Both kiosks have average monthly sales of 60 thousand MAD over these three months. An analyst proposes replacing the six records with those two averages for the following reports:

| Report | Purpose |
|---|---|
| A | Compare total sales contributed by each kiosk during January–March |
| B | Compare the kiosks' sales in March |
| C | Examine each kiosk's sales pattern when deciding April stock allocations |

a. For which reports would the two averages retain the information needed for the stated sales comparison? For the others, specify the records you would use and identify their data structure.

b. State what one observation represents in the original table and name the structure of that dataset.

c. Management proposes equal April stock allocations because the averages are equal. Give a recommendation supported by the available records.

---

### Question 3: Solution

a.

| Report | Appropriate records and assessment |
|---|---|
| A | The averages are sufficient to compare January–March sales totals because both cover the same three months. Equal averages imply equal totals here. The summary is a cross-sectional comparison of kiosks over a common period. |
| B | Retain Alpha and Beta's March records. These form a March cross-section. The two period averages do not preserve that month's comparison. |
| C | Retain January, February, and March for each kiosk. Each kiosk's records form a time series; keeping both series together gives panel data. |

b. One observation is a kiosk in a particular month. The original dataset is panel, or longitudinal, data because the same kiosks are observed repeatedly.

c. Equal April allocations are not justified by the averages alone. Alpha's sales rise from 40 to 80 thousand MAD, while Beta's fall from 80 to 40 thousand MAD.

Retain the monthly records and investigate current demand, product mix, stock already available, and any stock shortages before allocating stock. The patterns warrant different consideration, but three months of sales do not establish future demand or the appropriate allocation.

### Question 4 (Average)

Delta Transport wants to understand satisfaction among all individual passengers who used its intercity bus services during September.

| Survey information | Detail |
|---|---|
| Invitation list | 6,000 September passengers who booked through the mobile app |
| Invitation procedure | Every passenger on the list was invited once |
| Responses | 1,200 passengers, with one response per passenger |
| Result | 74% of respondents selected satisfied or very satisfied |
| Other booking channels | Website, ticket offices, and travel agents |

The survey team makes three statements:

1. Every passenger on our invitation list was invited.
2. We have a census of September passengers because every received response was analysed.
3. Passenger satisfaction in September was 74%.

a. Which statements can be retained in the public report, and which need revision? Give suitable wording where needed and identify the population actually described by the 74% figure.

b. Management offers funding to obtain twice as many responses next time. Recommend how the next survey should be organised and explain whether the larger response count would, by itself, support Statement 3.

---

### Question 4: Solution

a.

| Statement | Assessment |
|---|---|
| 1 | Retain. It accurately describes the invitation procedure, but says nothing about passengers outside the list or those who did not respond. |
| 2 | Revise. The target population is all individual September passengers. The 1,200 respondents are a sample, not a census of that population. |
| 3 | Revise to: “Among the 1,200 app-booking passengers who responded, 74% selected satisfied or very satisfied.” |

The 74% figure describes the respondent group. The invitation list excludes passengers who booked only through other channels; within the list, those who responded may differ from those who did not.

b. Build a passenger list covering all booking channels and remove duplicate passenger entries. Draw a random sample from that list, with coverage of the relevant routes and departure times. Use consistent questions and follow up with selected passengers who do not respond.

Doubling responses from the same restricted invitation method would not resolve the missing coverage or necessarily resolve response differences. Random selection supports broader inference, but coverage and nonresponse still need attention.

### Question 5 (Average)

Epsilon Retail compares two website checkout designs.

| Feature | Study A | Study B |
|---|---|---|
| Participants | 2,400 customers randomly selected from the current customer list; all participate | 200 loyalty-club members who volunteer |
| Design received | Each customer chooses standard or simplified checkout | Each participant is randomly assigned standard or simplified checkout |
| Prices and delivery terms | The same under both designs | The same under both designs |
| Period | Both designs operate during the same month | Both designs operate during the same month |
| Recorded outcome | Whether the customer completes a purchase | Whether the customer completes a purchase |

Each customer contributes one outcome. In both studies, the simplified checkout has the higher observed purchase-completion rate.

The marketing director prefers Study A because it includes more customers and they were randomly selected. The website manager prefers Study B because the design was randomly assigned and recommends applying its result to all customers.

a. Classify the studies and assess each manager's reasoning.

b. Which study would you rely on more for evidence that the design itself affects purchase completion? What limits remain when deciding whether to introduce it for all customers?

c. Describe a study that combines the useful features of these two designs.

---

### Question 5: Solution

a.

| Study | Classification | Strength and limitation |
|---|---|---|
| A | Observational: customers choose their design | Random selection supports representation of the current customer list. However, customers choosing the two designs may differ in purchase intention, experience, or device. Neither random selection nor the larger sample removes that problem. |
| B | Experimental: the design is randomly assigned | Random assignment strengthens the comparison between designs among participants. The volunteer loyalty-club sample does not establish representation of all customers. |

Each manager identifies a genuine strength but overlooks a different limitation.

b. Study B provides stronger causal evidence because the design is assigned rather than chosen. Random assignment reduces systematic differences between the groups receiving the two designs; it does not guarantee exact balance.

The observed difference may still reflect chance variation. Its size and uncertainty have not been reported. Loyalty-club volunteers may also respond differently from other customers. Neither study alone establishes the benefit of a general rollout.

c. Randomly select participants from a customer list covering the intended users, then randomly assign checkout designs within that sample. Keep the other relevant conditions comparable and record purchase completion consistently.

Selection addresses who is represented; assignment addresses the comparison between designs. Coverage, nonresponse, and changes in operating conditions still require attention.

### Question 6 (Hard)

Zeta Manufacturing uses scheduled machine inspections. When an inspection detects wear, components can be replaced before a breakdown occurs.

The company is considering a model that predicts whether a machine will break down in the following month.

| Project information | Detail |
|---|---|
| Records | 10,000 machine-months from two recently opened plants, collected under the existing inspection schedule |
| Other plants | Older plants keep paper maintenance records, which were not included |
| Model development | Twelve candidate models were developed and compared using the 10,000 records |
| Reported result | The selected model correctly classifies 96% of those same records |
| Outcomes in the reported data | 94% of machine-months had no breakdown |
| Exclusions before development | Some machine-months with confirmed breakdowns were labelled “exceptional operation” and left out; no further information about those cases is supplied |

Management proposes inspecting machines predicted to have no breakdown once a month instead of twice, at every plant. Machines predicted to break down would retain twice-monthly inspections.

Recommend whether and how the company should proceed. Support your recommendation with the report and specify what evidence would change your decision.

---

### Question 6: Solution

Retain the current inspection schedule while evaluating the model and the proposed policy. The report supports further investigation, not the proposed rollout.

| Evidence | Assessment |
|---|---|
| 96% accuracy; 94% of records have no breakdown | The model exceeds the accuracy of always predicting no breakdown. However, overall accuracy does not reveal how many breakdowns it misses or whether the improvement justifies changing inspections. |
| The same records are used for development, selection, and reported evaluation | Performance may be optimistic after selecting the best of twelve models. Independent evaluation is needed. |
| Some confirmed breakdown cases were excluded | Establish why they were excluded and whether those conditions can occur in the intended use. Genuine relevant breakdowns should not be removed simply because they are unusual. |
| Only two newer plants supply the records | Reliability at older plants has not been demonstrated. Their maintenance records need to be assessed and made comparable where possible. |
| Inspections can trigger replacement before failure | The recorded outcomes reflect the existing inspection schedule. Reducing inspections may increase breakdown risk, so accuracy under the old schedule does not establish performance under the proposed one. |

First evaluate the selected model on separate, previously unused records covering the intended plants and operating conditions. Report correct and missed breakdown predictions, false alarms, and performance relative to the no-breakdown benchmark.

If that evidence is satisfactory, test a limited policy change under a controlled pilot with safety safeguards. Compare breakdowns, downtime, maintenance costs, and inspection savings with a group retaining the current schedule.

Evidence of reliable prediction alone would not justify rollout. Evidence that the changed inspection policy maintains acceptable safety and operating outcomes at worthwhile cost would support proceeding.

### Question 12 (Average)

Mu Support retains only the following cumulative summary for twenty nonnegative customer waiting times. All waiting times were below 20 minutes.

| Waiting-time threshold | Number below the threshold |
|---|---:|
| 5 minutes | 4 |
| 10 minutes | 11 |
| 15 minutes | 17 |
| 20 minutes | 20 |

### Question 12: Solution

| Waiting time (minutes) | Frequency | Percent frequency |
|---|---:|---:|

### Question 15 (Hard)

Omicron Maintenance compares two repair teams. Every completed job is classified as routine or complex and as on time or late against its agreed deadline.

| Team | Job type | On-time jobs | Completed jobs |
|---|---|---:|---:|
| Alpha | Routine | 18 | 20 |
| Alpha | Complex | 56 | 80 |
| Beta | Routine | 68 | 80 |
| Beta | Complex | 12 | 20 |

The operations director recommends Beta for a new contract, citing its overall on-time record.

Evaluate the recommendation numerically. As part of your assessment, compare the teams on a reference workload of 20 routine and 20 complex jobs each, assuming their observed within-type on-time rates apply to that workload. Explain what the records support and what remains uncertain.

---

### Question 15: Solution

| On-time rate | Alpha | Beta |
|---|---:|---:|
| Routine | $18/20=90\%$ | $68/80=85\%$ |
| Complex | $56/80=70\%$ | $12/20=60\%$ |
| Overall | $(18+56)/(20+80)=74\%$ | $(68+12)/(80+20)=80\%$ |

Beta leads overall, but Alpha leads within both job types.

$$\text{Routine-job share: Alpha}=\frac{20}{100}=20\%,\qquad \text{Beta}=\frac{80}{100}=80\%$$

Beta's workload contains a much larger share of routine jobs, which have higher on-time rates for both teams. The aggregate ranking reverses the within-type ranking: Simpson's paradox.

For the common reference workload:

| Calculation | Alpha | Beta |
|---|---:|---:|
| Routine jobs on time | $20(0.90)=18$ | $20(0.85)=17$ |
| Complex jobs on time | $20(0.70)=14$ | $20(0.60)=12$ |
| Total on time | $18+14=32$ | $17+12=29$ |
| Overall rate | $32/40=80.0\%$ | $29/40=72.5\%$ |

Alpha has the stronger record for the common workload under the stated assumption. Beta's aggregate rate alone does not justify the recommendation.

These projections are not guaranteed outcomes or proof of a causal team effect. Job differences within each type, capacity, and whether the observed rates will apply to the new contract still need consideration.

### Question 17 (Easy)

Rho Repairs records the waiting times for eight customers.

| Customer | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 |

### Question 17: Solution

### Question 39 (Average)

Of the population share $0.0415$ receiving both flags, $0.032$ consists of defective components.

b. Exactly one test flags.

$$E=(F_1\cap F_2^c)\cup(F_1^c\cap F_2)$$

The two cases are mutually exclusive. Every probability in the following table is conditional on the status group.

| Test outcome | Defective ($D$) | Acceptable ($D^c$) |
|---|---:|---:|
| Test 1 flags; Test 2 does not | $0.80(0.20)=0.16$ | $0.10(0.90)=0.09$ |
| Test 1 does not flag; Test 2 flags | $0.20(0.80)=0.16$ | $0.90(0.10)=0.09$ |
| Exactly one flag within the group | $0.16+0.16=0.32$ | $0.09+0.09=0.18$ |

$$P(E\mid D)=P(F_1\mid D)P(F_2^c\mid D)+P(F_1^c\mid D)P(F_2\mid D)=0.32$$

$$P(E\mid D^c)=P(F_1\mid D^c)P(F_2^c\mid D^c)+P(F_1^c\mid D^c)P(F_2\mid D^c)=0.18$$

Multiply by the group shares to obtain joint probabilities in the full population:

| Actual status and exactly one flag | Joint probability |
|---|---:|
| Defective and exactly one flag: $D\cap E$ | $0.05(0.32)=0.016$ |
| Acceptable and exactly one flag: $D^c\cap E$ | $0.95(0.18)=0.171$ |
| Exactly one flag, either status: $E$ | $0.016+0.171=0.187$ |

$$P(E)=P(D)P(E\mid D)+P(D^c)P(E\mid D^c)=0.187$$

$$P(D\mid E)=\frac{P(D\cap E)}{P(E)}=\frac{P(D)P(E\mid D)}{P(E)}$$

$$P(D\mid E)=\frac{0.05(0.32)}{0.187}=\frac{16}{187}\approx0.0856$$

c. Independence in the full population.

$$P(F_i)=P(D)P(F_i\mid D)+P(D^c)P(F_i\mid D^c),\qquad i=1,2$$

$$P(F_1)=P(F_2)=0.05(0.80)+0.95(0.10)=0.135$$

| Independence check | Probability |
|---|---:|
| Both flags, calculated in part (a): $P(F_1\cap F_2)$ | $0.0415$ |
| Product required for independence: $P(F_1)P(F_2)$ | $0.135^2=0.018225$ |

$$P(F_1\cap F_2)=0.0415\ne0.018225=P(F_1)P(F_2)$$

Equivalently,

$$P(F_2\mid F_1)=\frac{P(F_1\cap F_2)}{P(F_1)}=\frac{0.0415}{0.135}\approx0.3074$$

$$P(F_2\mid F_1)\ne P(F_2)=0.135$$

The flags are not independent overall: observing the first flag raises the probability of a second flag from 13.5% to about 30.74%. Independence within status groups does not imply independence after the groups are combined.

### Question 49 (Hard)

Alpha Marketplace considers reviewing every order flagged by its fraud system.

| Quantity | Value |
|---|---:|
| Orders in each period | 10,000 |
| Fraud probability in the usual period | 0.03 |
| Fraud probability in a low-fraud period | 0.002 |
| Flag probability given fraud, in either period | 0.80 |
| Flag probability given a legitimate order, in either period | 0.05 |
| Review cost per flagged order | 12 MAD |
| Loss from each fraudulent order not reviewed | 150 MAD |

A review identifies the order's true status and prevents the entire fraud loss if the order is fraudulent. Legitimate orders cause no fraud loss. Unflagged orders are not reviewed, there is no review-capacity limit, and no other costs apply.

Recommend whether to review every flagged order in each period. Support the recommendation with expected counts and total costs compared with reviewing no orders. Also determine the review cost per flagged order at which the two policies have equal expected total cost in the usual period.

---

### Question 49: Solution

The counts below are expected counts under the stated probabilities, not guaranteed outcomes.

| Status and flag result | Usual period | Low-fraud period |
|---|---:|---:|
| Fraudulent and flagged | $10{,}000(0.03)(0.80)=240$ | $10{,}000(0.002)(0.80)=16$ |
| Fraudulent and unflagged | $300-240=60$ | $20-16=4$ |
| Legitimate and flagged | $9{,}700(0.05)=485$ | $9{,}980(0.05)=499$ |
| Legitimate and unflagged | $9{,}700-485=9{,}215$ | $9{,}980-499=9{,}481$ |
| Total flagged | $240+485=725$ | $16+499=515$ |
| Total fraudulent | 300 | 20 |

For the usual period:

$$C_{\text{review flags}}=725(12)+60(150)=8{,}700+9{,}000=17{,}700\text{ MAD}$$

$$C_{\text{review none}}=300(150)=45{,}000\text{ MAD}$$

$$\text{Expected saving}=45{,}000-17{,}700=27{,}300\text{ MAD}$$

For the low-fraud period:

$$C_{\text{review flags}}=515(12)+4(150)=6{,}180+600=6{,}780\text{ MAD}$$

$$C_{\text{review none}}=20(150)=3{,}000\text{ MAD}$$

$$\text{Extra expected cost}=6{,}780-3{,}000=3{,}780\text{ MAD}$$

Review every flag in the usual period, but not in the low-fraud period under the stated two policy choices and costs. The system's conditional flag probabilities have not changed; the fraud share among flags has:

$$P(\text{fraud}\mid\text{flag, usual})=\frac{240}{725}\approx0.3310$$

$$P(\text{fraud}\mid\text{flag, low-fraud})=\frac{16}{515}\approx0.0311$$

For the break-even review cost $c$ in the usual period:

$$725c+60(150)=300(150)$$

$$725c=(300-60)(150)=36{,}000$$

$$c=\frac{36{,}000}{725}\approx49.66\text{ MAD per flagged order}$$

The exact threshold is $36{,}000/725$. Review costs below it favour reviewing every flag; costs above it favour reviewing none.

### Question 50 (Hard)

Beta Distribution has two batches of distinct components.

| Batch | Defective | Acceptable | Total |
|---|---:|---:|---:|
| Alpha | 2 | 4 | 6 |
| Gamma | 1 | 3 | 4 |

Two sampling procedures are proposed:

* Procedure A: choose Alpha with probability 0.60 or Gamma with probability 0.40, then select two components uniformly without replacement from that chosen batch.
* Procedure B: pool all ten components, then select two uniformly without replacement from the pool.

Under Procedure A:

a. Calculate the probability that exactly one selected component is defective.

b. Given exactly one defect in the selected pair, calculate the probability that the pair came from Alpha.

c. Given that the selected pair contains at least one defect, calculate the probability that both components are defective.

Then calculate the probability of exactly one defect under Procedure B. A colleague argues that the procedures must give the same result because the batch-selection probabilities in A match the batches' shares of the ten components. Assess that argument.

---

### Question 50: Solution

Let $E$ denote exactly one defect, $H$ at least one defect, and $T$ two defects.

a. Within each batch, all unordered pairs are equally likely.

| Pair count | Alpha | Gamma |
|---|---:|---:|
| All pairs | $\binom{6}{2}=15$ | $\binom{4}{2}=6$ |
| Exactly one defect ($E$) | $\binom{2}{1}\binom{4}{1}=8$ | $\binom{1}{1}\binom{3}{1}=3$ |
| Two defects ($T$) | $\binom{2}{2}=1$ | $0$ |
| At least one defect ($H$) | $8+1=9$ | $3+0=3$ |

Divide each count by the total number of pairs in its batch:

| Conditional probability | Alpha | Gamma |
|---|---:|---:|
| Exactly one defect | $8/15$ | $3/6$ |
| Two defects | $1/15$ | $0/6=0$ |
| At least one defect | $9/15$ | $3/6$ |

$$P_A(E)=0.60\left(\frac{8}{15}\right)+0.40\left(\frac{3}{6}\right)=0.32+0.20=0.52$$

b.

$$P_A(\text{Alpha}\mid E)=\frac{P_A(\text{Alpha})P_A(E\mid\text{Alpha})}{P_A(E)}$$

$$P_A(\text{Alpha}\mid E)=\frac{0.60(8/15)}{0.52}=\frac{0.32}{0.52}=\frac{8}{13}\approx0.6154$$

c. Gamma cannot supply two defective components.

$$P_A(T)=0.60\frac{\binom{2}{2}}{\binom{6}{2}}+0.40(0)=0.60\left(\frac{1}{15}\right)=0.04$$

Exactly one defect and two defects are nonoverlapping and together make up at least one defect.

$$P_A(H)=P_A(E)+P_A(T)=0.52+0.04=0.56$$

Since $T\subseteq H$,

$$P_A(T\mid H)=\frac{P_A(T)}{P_A(H)}=\frac{0.04}{0.56}=\frac{1}{14}\approx0.0714$$

For Procedure B, the pool contains three defective and seven acceptable components.

$$P_B(E)=\frac{\binom{3}{1}\binom{7}{1}}{\binom{10}{2}}=\frac{3(7)}{45}=\frac{7}{15}\approx0.4667$$

The procedures are not equivalent. Procedure A always selects both components from the same batch. Procedure B also permits pairs drawn from different batches. Matching the batch shares for the initial selection does not make the possible pairs or their probabilities identical.
