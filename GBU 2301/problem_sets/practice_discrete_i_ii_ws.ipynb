### Question 1 (Easy)

Alpha Consulting randomly awards one of four contracts. Let $X$ be the fee earned.

| Contract | Beta | Gamma | Delta | Epsilon |
|---|---:|---:|---:|---:|
| Fee (MAD) | 12.50 | 20.00 | 27.50 | 50.00 |

Which statement is correct?

A. $X$ is continuous because some fees contain decimals.

B. $X$ is discrete because it has four possible values.

C. $X$ is continuous because money has a meaningful zero.

D. $X$ is discrete only if the four contracts are equally likely.

---

### Question 1: Solution

Answer: B.

The possible values are $12.50,\ 20,\ 27.50,\ 50$ MAD. A finite set is discrete, even when some values are not integers. Equal probabilities are not required.

Beta Repairs records the number of urgent jobs received on 20 comparable days.

Using these relative frequencies, what is the estimated probability of at least two urgent jobs on a comparable day?

A. 0.05

### Question 3 (Average)

Gamma Logistics uses the following distribution for a daily adjustment.

|---|---:|---:|---:|---:|

These are the only possible adjustments. What is the probability that the adjustment is zero or negative?

### Question 3: Solution

Answer: D.

Both the negative adjustment and the zero adjustment belong to the event.

### Question 4 (Easy)

Delta Retail offers four equally likely voucher values.

| Voucher value (MAD) | 0 | 10 | 30 | 40 |
|---|---:|---:|---:|---:|

Which statement correctly describes the expected voucher value?

A. It is 20 MAD, although no individual voucher has that value.

B. It is 20 MAD, so 20 MAD is the most likely voucher value.

C. It is 10 MAD because an expected value must be one of the possible values.

D. It is 40 MAD because repeated draws eventually produce the largest value.

This is the long-run average under the model, not a possible voucher value or a most likely outcome.

### Question 5 (Easy)

For Epsilon Services' daily number of cancellations, $E(X)=3$ and $E(X^2)=13$. What are the variance and standard deviation?

A. Variance 10; standard deviation $\sqrt{10}$

B. Variance 4; standard deviation 2

C. Variance 13; standard deviation $\sqrt{13}$

D. Variance 2; standard deviation 4

---

### Question 5: Solution

Answer: B.

$$\operatorname{Var}(X)=E(X^2)-[E(X)]^2=13-3^2=4,$$
$$\sigma_X=\sqrt4=2\text{ cancellations}.$$

Variance is measured in cancellations squared.

### Question 6 (Average)

Zeta Services starts with a daily allowance of 200 MAD. Each service call uses 30 MAD. If $X$ is the number of calls, the remaining balance is $Y=200-30X$.

| Measure of $X$ | Value |
|---|---:|
| Mean | 2 calls |
| Variance | 0.64 calls squared |

What are the mean and standard deviation of $Y$?

A. Mean 140 MAD; SD $-24$ MAD

B. Mean 140 MAD; SD 19.20 MAD

C. Mean 200 MAD; SD 24 MAD

D. Mean 140 MAD; SD 24 MAD

---

### Question 6: Solution

Answer: D.

$$E(Y)=200-30(2)=140\text{ MAD},$$
$$\operatorname{Var}(Y)=(-30)^2(0.64)=576\text{ MAD}^2,$$
$$\sigma_Y=\sqrt{576}=24\text{ MAD}.$$

A negative coefficient reverses the direction of change, not the sign of the standard deviation.

### Question 7 (Easy)

Eta Supplies has two pending orders. Let $X$ and $Y$ indicate whether the first and second orders are confirmed: 1 for confirmed, 0 otherwise. The table gives their joint probabilities.

| $X$ | $Y=0$ | $Y=1$ |
|---|---:|---:|

| 1 | 0.30 | 0.40 |

What is the probability that exactly one order is confirmed?

A. 0.10

B. 0.30

C. 0.40

D. 0.70

---

### Question 7: Solution

Answer: C.

$$P(X+Y=1)=P(X=0,Y=1)+P(X=1,Y=0)=0.10+0.30=0.40.$$

The two relevant pairs are mutually exclusive.

### Question 8 (Average)

Theta Transport records whether two vehicles arrive late. Each indicator is 1 for late and 0 for on time. The table gives their joint probabilities.

| First vehicle, $X$ | Second: $Y=0$ | Second: $Y=1$ |
|---|---:|---:|
| 0 | 0.10 | 0.20 |
| 1 | 0.30 | 0.40 |

Given that the first vehicle is late, what is the probability that the second is late?

A. 0.4000

B. 0.5714

C. 0.6000

D. 0.6667

---

### Question 8: Solution

Answer: B.

$$P(X=1)=0.30+0.40=0.70,$$
$$P(Y=1\mid X=1)=\frac{P(X=1,Y=1)}{P(X=1)}
=\frac{0.40}{0.70}\approx0.5714.$$

The denominator includes all outcomes with the first vehicle late.

### Question 10 (Average)

Kappa Retail's two outlets have daily sales counts $X$ and $Y$.

| Measure | Value |
|---|---:|
| $\operatorname{Var}(X)$ | 4 |
| $\operatorname{Var}(Y)$ | 9 |

Which pair is correct?

A. Covariance $-6$; correlation $-1$

B. Covariance 3; correlation 0.50

C. Covariance $-3$; correlation $-0.50$

D. Covariance 0; correlation 0

### Question 10: Solution

Answer: C.

### Question 11 (Hard)

Lambda Testing uses three possible paired outcomes.

| Outcome | Probability | $X$ | $Y$ |
|---|---:|---:|---:|
| A | 0.25 | $-2$ | 4 |
| B | 0.50 | 0 | 0 |
| C | 0.25 | 2 | 4 |

Which statement is correct?

A. $X$ and $Y$ are dependent and have positive covariance.

B. $X$ and $Y$ are independent and have zero covariance.

C. $X$ and $Y$ are dependent and have negative covariance.

D. $X$ and $Y$ are dependent and have zero covariance.

---

### Question 11: Solution

Answer: D.

$$E(X)=0,\qquad E(Y)=4(0.25)+0(0.50)+4(0.25)=2.$$

$$\operatorname{Cov}(X,Y)=(-2)(4-2)(0.25)+0(0-2)(0.50)+2(4-2)(0.25)
=-1+0+1=0.$$

However, $P(Y=0\mid X=0)=1$, while $P(Y=0)=0.50$. Knowing $X$ changes the distribution of $Y$, so the variables are dependent.

### Question 12 (Average)

Mu Investments places half its initial capital in each of two hypothetical funds.

| Measure | Fund Alpha | Fund Beta |
|---|---:|---:|
| Return SD (percentage points) | 6 | 4 |

The correlation of returns is zero. What is the portfolio's return standard deviation?

A. 5.000 percentage points

B. 3.606 percentage points

C. 13.000 percentage points

D. 2.000 percentage points

---

### Question 12: Solution

Answer: B.

Zero correlation gives zero covariance here.

$$\sigma_p^2=(0.50)^2(6^2)+(0.50)^2(4^2)+2(0.50)(0.50)(0)
=9+4=13.$$

$$\sigma_p=\sqrt{13}\approx3.606\text{ percentage points}.$$

The average of the individual standard deviations is not the portfolio standard deviation.

### Question 13 (Easy)

Nu Couriers' contract requires the company to pay an additional fixed daily charge of 40 MAD. The distribution of deliveries and all other revenues and costs remain unchanged.

How does this affect the distribution of daily net profit?

A. Both its mean and standard deviation fall by 40 MAD.

B. Its mean is unchanged, but its standard deviation falls by 40 MAD.

C. Its mean falls by 40 MAD and its variance falls by $40^2$.

D. Its mean falls by 40 MAD and its standard deviation is unchanged.

---

### Question 13: Solution

Answer: D.

If original profit is $Y$, new profit is $Z=Y-40$.

$$E(Z)=E(Y)-40,\qquad \operatorname{Var}(Z)=\operatorname{Var}(Y).$$

The fixed charge shifts all outcomes by the same amount without changing their deviations from the mean.

### Question 14 (Average)

Xi Services has two probability models for daily demands $X$ and $Y$. Both models give the same marginal distribution for each demand, but different joint distributions.

Which result must be the same in both models?

A. The expected combined demand, $E(X+Y)$

B. The variance of combined demand, $\operatorname{Var}(X+Y)$

C. The probability of meeting both demands with a shared capacity

D. The correlation between the two demands

---

### Question 14: Solution

Answer: A.

$$E(X+Y)=E(X)+E(Y).$$

The marginal distributions determine these two means. The joint distribution affects covariance, the variance of the total, and probabilities involving both demands.

### Question 15 (Average)

Omicron Investments uses the following one-year return scenarios. Portfolio weights are nonnegative and sum to one.

| State | Probability | Alpha return (%) | Beta return (%) |
|---|---:|---:|---:|
| Weak | 0.50 | 0 | 12 |
| Strong | 0.50 | 12 | 4 |

Which allocation gives an expected return of 7%, and what is its return standard deviation?

A. 25% Alpha, 75% Beta; SD 1 percentage point

B. 50% Alpha, 50% Beta; SD 5 percentage points

C. 50% Alpha, 50% Beta; SD 1 percentage point

D. 75% Alpha, 25% Beta; SD 5 percentage points

---

### Question 15: Solution

Answer: C.

$$E(R_A)=6\%,\qquad E(R_B)=8\%.$$

For Alpha weight $w$,

$$6w+8(1-w)=7\quad\Rightarrow\quad w=0.50.$$

The portfolio returns are $0.50(0)+0.50(12)=6\%$ and $0.50(12)+0.50(4)=8\%$.

Variance is in squared percentage points; standard deviation is in percentage points.

$$\sigma_p^2=(6-7)^2(0.50)+(8-7)^2(0.50)=1,\qquad \sigma_p=1.$$

### Question 16 (Easy)

Pi Repairs records urgent jobs on 40 comparable days.

| Urgent jobs, $x$ | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| Number of days | 6 | 14 | 12 | 8 |

a. Construct the empirical probability distribution. Calculate its mean, variance and standard deviation.

b. If the same distribution applies on each of the next 100 days, how many of those days are expected to have at least two urgent jobs?

c. The team can complete two urgent jobs per day. Additional jobs are outsourced. Find the probability of outsourcing and the expected number outsourced per day.

---

### Question 16: Solution

a.

| $x$ | $f(x)$ | $xf(x)$ | $(x-1.55)^2f(x)$ |
|---|---:|---:|---:|
| 0 | $6/40=0.15$ | 0.00 | $(-1.55)^2(0.15)=0.360375$ |
| 1 | $14/40=0.35$ | 0.35 | $(-0.55)^2(0.35)=0.105875$ |
| 2 | $12/40=0.30$ | 0.60 | $0.45^2(0.30)=0.060750$ |
| 3 | $8/40=0.20$ | 0.60 | $1.45^2(0.20)=0.420500$ |
| Total | 1.00 | 1.55 | 0.947500 |

$$E(X)=1.55\text{ jobs},\qquad
\operatorname{Var}(X)=0.9475\text{ jobs}^2,\qquad
\sigma_X=\sqrt{0.9475}\approx0.9734\text{ jobs}.$$

b.

$$P(X\geq2)=0.30+0.20=0.50,$$
$$E(\text{days with at least two jobs})=100(0.50)=50\text{ days}.$$

This is an expected count, not a guaranteed number of days.

c. Outsourcing occurs only when $X=3$; exactly one job is then outsourced.

$$P(\text{outsourcing})=0.20,$$
$$E(\text{jobs outsourced})=0(0.80)+1(0.20)=0.20\text{ jobs per day}.$$

### Question 17 (Average)

Rho Deliveries receives a daily service adjustment.

| Adjustment, $X$ (MAD) | $-30$ | 0 | 60 | 120 |
|---|---:|---:|---:|---:|
| Probability | $a$ | 0.20 | $2a$ | 0.20 |

These are the only outcomes.

a. Find $a$. Calculate the expected adjustment and its standard deviation.

b. Find the probability that the adjustment is below its expected value.

c. A new fixed daily fee $F\geq0$ will be deducted from the adjustment. Find the greatest fee that leaves a nonnegative net adjustment on at least 60% of days. Breaking even counts as nonnegative.

---

### Question 17: Solution

a.

$$a+0.20+2a+0.20=1\quad\Rightarrow\quad a=0.20.$$

| $x$ (MAD) | $f(x)$ | $xf(x)$ | $x^2f(x)$ |
|---|---:|---:|---:|
| $-30$ | 0.20 | $-6$ | 180 |
| 0 | 0.20 | 0 | 0 |
| 60 | 0.40 | 24 | 1,440 |
| 120 | 0.20 | 24 | 2,880 |
| Total | 1.00 | 42 | 4,500 |

$$E(X)=42\text{ MAD},$$
$$\operatorname{Var}(X)=4{,}500-42^2=2{,}736\text{ MAD}^2,$$
$$\sigma_X=\sqrt{2{,}736}\approx52.31\text{ MAD}.$$

b.

$$P(X<E(X))=P(X<42)=P(X=-30)+P(X=0)=0.40.$$

c. Nonnegative net adjustment requires $X-F\geq0$, or $X\geq F$.

| Fee | Outcomes that cover the fee | Coverage probability |
|---|---|---:|
| $F=0$ | 0, 60, 120 | 0.80 |
| $0<F\leq60$ | 60, 120 | 0.60 |
| $60<F\leq120$ | 120 | 0.20 |
| $F>120$ | None | 0.00 |

The greatest qualifying fee is 60 MAD. At that fee, an adjustment of 60 MAD gives exactly zero net adjustment and is included.

### Question 19 (Average)

Tau Retail can choose either payment arrangement below. The choice does not change the sales distribution. The entries are net daily profits after all costs.

| Units sold, $x$ | Probability | Arrangement A (MAD) | Arrangement B (MAD) |
|---|---:|---:|---:|
| 0 | 0.25 | $-40$ | 10 |
| 1 | 0.50 | 40 | 50 |
| 2 | 0.25 | 120 | 90 |

a. Calculate each arrangement's expected profit and standard deviation.

b. The owner requires expected profit of at least 45 MAD and a profit standard deviation no greater than 35 MAD. Which arrangement qualifies?

c. Which arrangement gives the greater probability of earning at least 100 MAD in a day? Explain why this comparison need not produce the same choice as part (b).

---

### Question 19: Solution

a.

$$E(Y_A)=(-40)(0.25)+40(0.50)+120(0.25)=40\text{ MAD},$$
$$E(Y_B)=10(0.25)+50(0.50)+90(0.25)=50\text{ MAD}.$$

| $x$ | Probability | $(Y_A-40)^2p$ | $(Y_B-50)^2p$ |

| 0 | 0.25 | $(-80)^2(0.25)=1{,}600$ | $(-40)^2(0.25)=400$ |

|---|---|---|---|

### Question 20 (Average)

The table gives the joint probabilities of order counts at Upsilon Supplies' two outlets on the same day.

| First outlet, $X$ | Second outlet: $Y=0$ | Second outlet: $Y=1$ |
|---|---:|---:|
| 0 | 0.10 | 0.20 |
| 1 | 0.25 | 0.15 |
| 2 | 0.20 | 0.10 |

a. Obtain both marginal distributions. Calculate the mean and variance of each order count.

b. Find $P(Y=1\mid X=2)$. Are the order counts independent?

c. Find the distribution and variance of the total number of orders, $S=X+Y$.

---

### Question 20: Solution

a.

| Orders | $f_X(x)$ | $f_Y(y)$ |
|---|---:|---:|
| 0 | $0.10+0.20=0.30$ | $0.10+0.25+0.20=0.55$ |
| 1 | $0.25+0.15=0.40$ | $0.20+0.15+0.10=0.45$ |
| 2 | $0.20+0.10=0.30$ | 0.00 |

$$E(X)=0(0.30)+1(0.40)+2(0.30)=1,$$
$$\operatorname{Var}(X)=(-1)^2(0.30)+0^2(0.40)+1^2(0.30)=0.60.$$

$$E(Y)=0(0.55)+1(0.45)=0.45,$$
$$\operatorname{Var}(Y)=(-0.45)^2(0.55)+0.55^2(0.45)=0.2475.$$

Means are in orders; variances are in orders squared.

b.

$$P(Y=1\mid X=2)=\frac{0.10}{0.20+0.10}=\frac13\ne0.45=P(Y=1).$$

The counts are not independent.

c.

| $s$ | Pairs $(x,y)$ | $P(S=s)$ | $(s-1.45)^2P(S=s)$ |
|---|---|---:|---:|
| 0 | $(0,0)$ | 0.10 | 0.210250 |
| 1 | $(0,1),(1,0)$ | $0.20+0.25=0.45$ | 0.091125 |
| 2 | $(1,1),(2,0)$ | $0.15+0.20=0.35$ | 0.105875 |
| 3 | $(2,1)$ | 0.10 | 0.240250 |
| Total | | 1.00 | 0.647500 |

$$E(S)=0(0.10)+1(0.45)+2(0.35)+3(0.10)=1.45\text{ orders},$$
$$\operatorname{Var}(S)=0.6475\text{ orders}^2.$$

### Question 21 (Average)

Phi Couriers uses three possible demand scenarios for two routes.

| Scenario | Probability | Route A orders, $X$ | Route B orders, $Y$ |
|---|---:|---:|---:|
| A | 0.25 | 0 | 2 |
| B | 0.50 | 2 | 1 |
| C | 0.25 | 4 | 0 |

a. Calculate both means and variances, covariance, and correlation.

b. Find the variance of total orders $X+Y$, first from the total-order distribution and then using covariance.

c. Does the correlation imply that total orders are constant? Support your answer numerically.

---

### Question 21: Solution

a.

$$\mu_X=0(0.25)+2(0.50)+4(0.25)=2,\qquad
\mu_Y=2(0.25)+1(0.50)+0(0.25)=1.$$

| Scenario | $p$ | $x-2$ | $y-1$ | $(x-2)^2p$ | $(y-1)^2p$ | $(x-2)(y-1)p$ |
|---|---:|---:|---:|---:|---:|---:|
| A | 0.25 | $-2$ | 1 | 1 | 0.25 | $-0.50$ |
| B | 0.50 | 0 | 0 | 0 | 0 | 0 |
| C | 0.25 | 2 | $-1$ | 1 | 0.25 | $-0.50$ |
| Total | 1.00 | | | 2 | 0.50 | $-1.00$ |

$$\operatorname{Var}(X)=2,\quad \operatorname{Var}(Y)=0.50,\quad
\operatorname{Cov}(X,Y)=-1,$$
$$\rho_{XY}=\frac{-1}{\sqrt2\sqrt{0.50}}=-1.$$

Means are in orders; variances and covariance are in orders squared. Correlation is unitless.

b.

| Total orders, $s$ | Probability | $(s-3)^2P(S=s)$ |
|---|---:|---:|
| 2 | 0.25 | 0.25 |
| 3 | 0.50 | 0 |
| 4 | 0.25 | 0.25 |

$$\operatorname{Var}(S)=0.25+0+0.25=0.50\text{ orders}^2.$$

Using covariance gives

$$\operatorname{Var}(X+Y)=2+0.50+2(-1)=0.50\text{ orders}^2.$$

c. No. The total takes values 2, 3 and 4, with positive variance. Here $Y=2-X/2$: a two-order rise on A accompanies only a one-order fall on B, so the changes do not cancel in $X+Y$.

### Question 22 (Average)

Chi Repairs' daily contribution before a two-day fixed expense has the distribution below. Contributions on the two days are independent.

| Daily contribution (MAD) | 0 | 40 | 100 |
|---|---:|---:|---:|
| Probability | 0.20 | 0.50 | 0.30 |

a. Construct the distribution of total contribution over two days. Show how repeated totals are combined.

b. Calculate the mean and standard deviation of the two-day total.

c. A fixed expense of 90 MAD must be covered by the two-day total. What is the probability of full coverage?

---

### Question 22: Solution

a. Independence gives each pair probability as the product of its daily probabilities.

| First day | Second day | Total, $s$ (MAD) | Joint probability |
|---:|---:|---:|---:|
| 0 | 0 | 0 | $0.20(0.20)=0.04$ |
| 0 | 40 | 40 | $0.20(0.50)=0.10$ |
| 40 | 0 | 40 | $0.50(0.20)=0.10$ |
| 40 | 40 | 80 | $0.50(0.50)=0.25$ |
| 0 | 100 | 100 | $0.20(0.30)=0.06$ |
| 100 | 0 | 100 | $0.30(0.20)=0.06$ |
| 40 | 100 | 140 | $0.50(0.30)=0.15$ |
| 100 | 40 | 140 | $0.30(0.50)=0.15$ |
| 100 | 100 | 200 | $0.30(0.30)=0.09$ |

| Total, $s$ (MAD) | 0 | 40 | 80 | 100 | 140 | 200 |
|---|---:|---:|---:|---:|---:|---:|
| $P(S=s)$ | 0.04 | 0.20 | 0.25 | 0.12 | 0.30 | 0.09 |

b. For one day,

$$E(X)=0(0.20)+40(0.50)+100(0.30)=50\text{ MAD},$$
$$E(X^2)=40^2(0.50)+100^2(0.30)=3{,}800\text{ MAD}^2,$$
$$\operatorname{Var}(X)=3{,}800-50^2=1{,}300\text{ MAD}^2.$$

For the independent days,

### Question 46 (Hard)

Psi Reservations has one operating condition throughout a randomly selected 20-minute window. The condition is initially unknown. Conditional on that condition, arrivals follow a Poisson process at the stated rate.

| Condition | Probability | Arrival rate |
|---|---:|---:|
| Regular | 0.60 | 3 calls per hour |
| Busy | 0.40 | 12 calls per hour |

a. Find the probability of exactly two calls in the window.

b. The window contains exactly two calls. Find the probability that the condition was busy. Does this evidence increase or decrease the busy-condition probability?

c. A single Poisson model is fitted using the probability-weighted average rate. Calculate its probability of exactly two calls and compare it with part (a).

---

### Question 46: Solution

a. Let $A=\{X=2\}$, and let $R$ and $B$ denote regular and busy conditions.

$$\mu_R=3(20/60)=1,\qquad \mu_B=12(20/60)=4.$$

| Condition | Prior probability | Conditional probability of two calls | Joint probability: condition and two calls |
|---|---:|---|---|
| Regular | 0.60 | $P(A\mid R)=e^{-1}/2\approx0.183940$ | $0.60P(A\mid R)\approx0.110364$ |
| Busy | 0.40 | $P(A\mid B)=4^2e^{-4}/2\approx0.146525$ | $0.40P(A\mid B)\approx0.058610$ |
| Total | 1.00 | | $P(A)\approx0.168974$ |

$$P(A)=0.60\left(\frac{e^{-1}}2\right)+0.40\left(8e^{-4}\right)
\approx0.168974.$$

b. Among windows with two calls, the busy share is

$$P(B\mid A)=\frac{P(B\cap A)}{P(A)}
=\frac{0.40(8e^{-4})}{0.60(e^{-1}/2)+0.40(8e^{-4})}
\approx0.346859.$$

The probability falls from 40% to about 34.69%. Exactly two calls are more likely under the regular condition than under the busy condition.

c.

$$\bar\lambda=0.60(3)+0.40(12)=6.60\text{ calls per hour},$$
$$\bar\mu=6.60(20/60)=2.20.$$

The single Poisson model gives

$$P(X=2)=\frac{2.20^2e^{-2.20}}{2!}\approx0.268144,$$

rather than 0.168974. Both models have expected count 2.20, but averaging rates does not produce the actual mixture of probabilities.

### Question 47 (Average)

Omega Services selects three different employees from a team of nine. Four are certified. Every group of three is equally likely.

a. Construct the complete probability distribution of the number of certified employees selected.

b. Calculate its mean, variance and standard deviation.

c. Find the probability that the group contains both certified and non-certified employees.

---

### Question 47: Solution

a. Let $X$ count certified employees. The parameters are $N=9$, $r=4$, $n=3$, with possible counts 0, 1, 2, 3.

$$P(X=x)=\frac{\binom4x\binom5{3-x}}{\binom93},
\qquad \binom93=84.$$

| $x$ | Certified choices | Non-certified choices | $P(X=x)$ |
|---|---:|---:|---:|
| 0 | $\binom40=1$ | $\binom53=10$ | $10/84$ |
| 1 | $\binom41=4$ | $\binom52=10$ | $40/84$ |
| 2 | $\binom42=6$ | $\binom51=5$ | $30/84$ |
| 3 | $\binom43=4$ | $\binom50=1$ | $4/84$ |
| Total | | | $84/84=1$ |

b.

$$E(X)=3\left(\frac49\right)=\frac43\approx1.3333\text{ employees},$$
$$\operatorname{Var}(X)=3\left(\frac49\right)\left(\frac59\right)
\left(\frac{9-3}{9-1}\right)=\frac59\text{ employees}^2,$$
$$\sigma=\sqrt{\frac59}\approx0.7454\text{ employees}.$$

c. Both categories are represented when $X=1$ or $X=2$.

$$P(1\leq X\leq2)=\frac{40+30}{84}=\frac56\approx0.8333.$$

### Question 48 (Average)

Beta Testing compares two random sampling procedures for a lot of eight items, exactly three of which are defective.

| Procedure | Sampling rule |
|---|---|
| A | Draw four different items without replacement. |
| B | Draw four times, replacing the selected item and mixing the lot after every draw. |

For B, draws are independent. A defective item may be selected more than once and counts on each draw.

a. State the exact model and possible defect counts for each procedure.

b. Calculate the mean and variance under each procedure.

c. Find the probability of at least three recorded defects under each procedure. Is the binomial model a good substitute for A for this event?

---

### Question 48: Solution

a.

| Procedure | Exact model | Possible counts |
|---|---|---|
| A | Hypergeometric: $N=8,\ r=3,\ n=4$ | 0, 1, 2, 3 |
| B | Binomial: $n=4,\ p=3/8$ | 0, 1, 2, 3, 4 |

A cannot contain four defects because only three defective items exist. B can record the same defective item repeatedly.

b. Both means are

$$E(X)=4\left(\frac38\right)=1.50\text{ recorded defects}.$$

Variances are in recorded defects squared.

For B,

$$\operatorname{Var}(X_B)=4\left(\frac38\right)\left(\frac58\right)=\frac{15}{16}=0.9375.$$

For A,

$$\operatorname{Var}(X_A)=\frac{15}{16}\left(\frac{8-4}{8-1}\right)
=\frac{15}{28}\approx0.535714.$$

c. Under A, at least three means exactly three.

$$P(X_A\geq3)=\frac{\binom33\binom51}{\binom84}
=\frac{1(5)}{70}=\frac1{14}\approx0.071429.$$

Under B,

$$P(X_B=3)=4\left(\frac38\right)^3\left(\frac58\right)
=\frac{540}{4096},$$
$$P(X_B=4)=\left(\frac38\right)^4=\frac{81}{4096},$$
$$P(X_B\geq3)=\frac{621}{4096}\approx0.151611.$$

The binomial probability is more than twice the exact probability for A, a difference of about 8.02 percentage points. Sampling half the lot creates substantial dependence, so this is not a close approximation for the event.

### Question 49 (Hard)

Gamma Imports receives a lot of ten items containing exactly three defective items. Tests identify defects perfectly.

The inspection rule is:

* Randomly select and inspect two different items.
* If either is defective, reject the lot and stop.
* Otherwise, randomly select and inspect two more items from those remaining.
* Accept the lot only if all four inspected items are nondefective.

Each item inspected costs 3 MAD. Both items in a stage are tested before the decision for that stage. There are no other inspection costs.

a. Find the probability of reaching the second stage and the probability of eventual acceptance.

b. Calculate the expected number of items inspected and expected inspection cost.

c. Compare this rule with always inspecting four different items and accepting only if all four are nondefective. Does stopping early change the acceptance probability or expected cost?

---

### Question 49: Solution

a. Initially, seven of the ten items are nondefective. Reaching stage two requires two nondefective items in stage one.

$$P(\text{stage 2})=\frac{\binom72}{\binom{10}2}
=\frac{21}{45}=\frac7{15}.$$

After two nondefective items have been removed, eight remain: five nondefective and three defective.

$$P(\text{second pair clean}\mid\text{first pair clean})
=\frac{\binom52}{\binom82}=\frac{10}{28}=\frac5{14}.$$

Therefore,

$$P(\text{accept})=\frac7{15}\frac5{14}=\frac16\approx0.166667.$$

b.

| Inspection path | Probability | Items inspected | Cost (MAD) |
|---|---:|---:|---:|
| Stop after stage one | $1-7/15=8/15$ | 2 | 6 |
| Reach stage two | $7/15$ | 4 | 12 |

$$E(\text{items inspected})=2\left(\frac8{15}\right)+4\left(\frac7{15}\right)
=\frac{44}{15}\approx2.9333\text{ items},$$
$$E(C)=6\left(\frac8{15}\right)+12\left(\frac7{15}\right)
=\frac{132}{15}=8.80\text{ MAD}.$$

c. With four inspections on every lot,

$$P(\text{accept})=\frac{\binom74}{\binom{10}4}
=\frac{35}{210}=\frac16.$$

| Rule | Acceptance probability | Expected inspection cost |
|---|---:|---:|
| Two-stage stopping rule | $1/6$ | 8.80 MAD |
| Always inspect four | $1/6$ | 12.00 MAD |

Early stopping saves $12-8.80=3.20$ MAD per lot on average without changing acceptance probability. Any defect found in the first pair already makes acceptance impossible under either rule.

### Question 50 (Hard)

Delta Rentals randomly selects three different devices from eight in storage. Exactly two of the eight are faulty. Every group of three is equally likely. Faulty devices are identified after selection and cannot be rented out.

Three customers each decide independently whether to rent one device, with rental probability 0.50. Customer decisions are independent of the device selection.

Before the selection and customer decisions, Delta must choose whether to reserve one additional device that is guaranteed to work.

| Financial item | Amount |
|---|---:|
| Contribution per device rented, after rental-related costs | 40 MAD |
| Fixed reservation fee for the additional device | 8 MAD, paid even if unused |
| Other relevant costs | 0 MAD |

Each available device can serve one customer. Unmet demand is lost, with no further charge.

a. Find the distributions of usable devices in the original three-device selection and customer demand.

b. For each reservation choice, calculate expected profit and the probability of meeting all customer demand. Recommend the higher-expected-profit option among those meeting all demand with probability at least 95%.

---

### Question 50: Solution

a. Let $X$ count usable devices in the original selection. Six of the eight work, so $X$ is hypergeometric with $N=8$, $r=6$, $n=3$.

$$P(X=x)=\frac{\binom6x\binom2{3-x}}{\binom83},\qquad \binom83=56.$$

| Usable devices, $x$ | Calculation | $P(X=x)$ |
|---|---|---:|
| 1 | $\binom61\binom22/56$ | $6/56$ |
| 2 | $\binom62\binom21/56$ | $30/56$ |
| 3 | $\binom63\binom20/56$ | $20/56$ |

Let $D$ count customers who rent. Then $D\sim\operatorname{Bin}(3,0.50)$.

| Demand, $d$ | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| $P(D=d)$ | $1/8$ | $3/8$ | $3/8$ | $1/8$ |

b. Without the reservation, rentals equal $\min(D,X)$. With it, rentals equal $\min(D,X+1)$.

For example, with one usable original device and no reservation:

$$E(\text{rentals}\mid X=1)=0\left(\frac18\right)
+1\left(\frac38+\frac38+\frac18\right)=\frac78=0.875,$$
$$P(\text{all demand met}\mid X=1)=P(D\leq1)=\frac18+\frac38=0.50.$$

The same calculation for each capacity gives:

| Original usable devices | $P(X=x)$ | Expected rentals: no reservation | Expected rentals: reserve | Service probability: no reservation | Service probability: reserve |
|---:|---:|---:|---:|---:|---:|
| 1 | $6/56$ | 0.875 | 1.375 | 0.500 | 0.875 |
| 2 | $30/56$ | 1.375 | 1.500 | 0.875 | 1.000 |
| 3 | $20/56$ | 1.500 | 1.500 | 1.000 | 1.000 |

For capacity two, for example, expected rentals are
$0(1/8)+1(3/8)+2(3/8)+2(1/8)=1.375$.

Weight the conditional results by the device-selection probabilities:

$$E(\text{rentals without reservation})
=\frac6{56}(0.875)+\frac{30}{56}(1.375)+\frac{20}{56}(1.50)
\approx1.366071,$$
$$E(\text{rentals with reservation})
=\frac6{56}(1.375)+\frac{30}{56}(1.50)+\frac{20}{56}(1.50)
\approx1.486607.$$

$$P(\text{all demand met, no reservation})
=\frac6{56}(0.50)+\frac{30}{56}(0.875)+\frac{20}{56}(1)
\approx0.879464,$$
$$P(\text{all demand met, reserve})
=\frac6{56}(0.875)+\frac{30}{56}(1)+\frac{20}{56}(1)
\approx0.986607.$$

Using unrounded expected rentals:

$$E(\pi_{\text{no reservation}})=40(1.366071\ldots)\approx54.64\text{ MAD},$$
$$E(\pi_{\text{reserve}})=40(1.486607\ldots)-8\approx51.46\text{ MAD}.$$

| Choice | Expected profit (MAD) | Probability all demand is met | Meets 95%? |
|---|---:|---:|---|
| No reservation | 54.64 | 87.95% | No |
| Reserve one device | 51.46 | 98.66% | Yes |

Reserve the additional device. It is the only qualifying choice, although expected profit is about 3.18 MAD lower. Without the service requirement, no reservation would maximize expected profit.
