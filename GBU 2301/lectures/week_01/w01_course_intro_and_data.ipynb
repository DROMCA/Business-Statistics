### Data and Data Sets

* Data: facts and figures collected for analysis and interpretation.

* Data set: all data collected for a particular study.

* Element: the entity on which data are collected.

* Variable: a characteristic recorded for each element.

* Value: the recorded measurement or label for one variable.

* Observation: the complete set of recorded values for one element, usually one row of a rectangular data table.

For a complete rectangular data set with n observations and p variables, the table contains n × p recorded values, excluding headers.

### Categorical and Quantitative Data

Categorical data identify a quality or group. Their values may be names or numeric codes.

* channel, category, and returning_customer are categorical.

* transaction_id is an identifier and is usually treated as categorical.

Quantitative data indicate how much or how many. Arithmetic differences have a meaningful interpretation.

* wait_minutes and amount_mad are quantitative.

* Quantitative data may be discrete or continuous.

    * Discrete data count how many, such as the number of products in a transaction.

    * Continuous data measure how much, such as wait_minutes.

Satisfaction uses numbers, but those numbers primarily indicate ordered categories. Numeric appearance alone does not determine statistical meaning.

---

### Scales of Measurement

| Scale | Information carried by the values | Meaningful comparisons |
|---|---|---|
| Nominal | Category identity | Same or different |
| Ordinal | Category identity and order | Higher or lower |
| Interval | Order and meaningful differences | Differences |
| Ratio | Order, differences, and a meaningful zero | Differences and ratios |

The scale of measurement affects which summaries, graphs, and statistical methods are appropriate.

---

### Nominal and Ordinal Scales

#### Nominal Scale

* Values distinguish categories without ranking them.

* Examples: ordering channel, product category, nationality, and account type.

* A numeric category code does not create quantitative data.

#### Ordinal Scale

* Values distinguish and rank categories.

* Examples: credit rating, class standing, and satisfaction category.

* The distance between adjacent categories is not necessarily constant.

---

### Interval and Ratio Scales

#### Interval Scale

* Differences between values are meaningful.

* Zero is a reference point rather than the absence of the measured quantity.

* Examples: temperature in Celsius and calendar year.

#### Ratio Scale

* Differences and ratios are meaningful.

* Zero indicates the absence of the measured quantity.

* Examples: waiting time, transaction amount, revenue, age, and units sold.

*A waiting time of 20 minutes is twice as long as 10 minutes. A temperature of 20°C is not twice as hot as 10°C because 0°C does not represent the absence of temperature.*

---

### Cross-Sectional and Time-Series Data

Cross-sectional data describe different elements at the same or approximately the same point in time.

* Revenue for 40 branches during August 2026

* Credit ratings for 60 companies on the same date

Time-series data describe the same variable over several time periods.

* Monthly revenue for one branch from 2024 through 2026

* Daily exchange rates during one year

When the same branches are observed repeatedly over time, the complete structure is panel data, also called longitudinal data. 

- Panel data combine cross-sectional and time-series dimensions.

### Data Sources

#### Existing Sources

* Internal records: sales, accounting, inventory, employees, customers, and operations

* External sources: government agencies, industry associations, commercial databases, and research publications

#### New Data Collection

* Surveys and interviews

* Direct observation

* Experiments

Availability does not establish suitability. Data collected for another purpose may use a different population, definition, frequency, or unit of analysis.

---

### Observational and Experimental Studies

In an observational study, the analyst observes variables without assigning the conditions being compared.

* Customers choose mobile or in-store ordering.

* Firms choose whether and when to adopt a technology.

In an experimental study, the researcher assigns one or more conditions and observes the resulting outcomes.

* Comparable customers are randomly assigned different website designs.

* Similar production periods are assigned different operating procedures.

Observational data can reveal important associations. A credible causal conclusion requires stronger design and careful treatment of alternative explanations.

---

### Data-Acquisition Considerations

* Relevance: Were the data collected for the current question?

* Definition: Are variables measured consistently across elements and periods?

* Coverage: Which elements, locations, or periods are absent?

* Accuracy: Can recording, coding, or processing errors be detected?

* Timeliness: Are the data still useful for the decision?

* Cost: Is additional information worth the time and expense?

* Access and responsibility: Is the collection and use of the data permitted and appropriate?

**The quality of an analysis cannot exceed the quality and relevance of the data used.**

### Descriptive Statistics

Descriptive statistics summarize the data that were observed.

Common forms include:

* Tabular summaries

* Graphical summaries

* Numerical summaries

Examples from the café data include the number of recorded transactions, the distribution of product categories, and the mean waiting time in the observed rows.

A descriptive result may be accurate for the recorded data while still being unrepresentative of a broader population.

---

### Tabular Summary Example

A frequency table summarizes the internal credit ratings of the 40 branches. Since the rating repeats every month, use one row per branch before counting.

---

```python
bank = pd.read_excel("../../datasets/BankBranchMonthly.xlsx", sheet_name="BranchMonthly")
branch_profiles = bank.drop_duplicates("branch_id")
rating_order = ["AAA", "AA", "A", "BBB", "BB"]

rating_summary = (
    branch_profiles["internal_credit_rating"]
    .value_counts()
    .reindex(rating_order, fill_value=0)
    .rename_axis("internal_credit_rating")
    .reset_index(name="branches")
)
rating_summary["percent_of_branches"] = (
    100 * rating_summary["branches"] / rating_summary["branches"].sum()
).round(1)

display(rating_summary)
```

### Graphical Summary Example

Aggregate the branch-month observations by month to examine total revenue over time.

---

```python
import matplotlib.pyplot as plt

monthly_revenue = (
    bank.groupby("month", as_index=False)["monthly_revenue_mad"]
    .sum()
)

plt.figure(figsize=(10, 4.8))
plt.plot(
    monthly_revenue["month"],
    monthly_revenue["monthly_revenue_mad"] / 1_000_000,
    color="#1f6f8b",
    marker="o",
    linewidth=2,
)
plt.title("Monthly Revenue Across 40 Branches")
plt.xlabel("Month")
plt.ylabel("Revenue (million MAD)")
plt.grid(alpha=0.25)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Numerical Summary Example

### Population, Sample, Census, and Sample Survey

* Population: the complete set of elements relevant to a study.

* Sample: a subset of the population from which data are collected.

* Census: data collection from every element in the population.

* Sample survey: data collection from a sample of the population.

If the café manager wants to understand all Fall 2026 lunch transactions, those transactions form the population. The 12 classroom rows serve as a hypothetical sample for illustration; they were not collected from that population.

A population is defined by the question, including the relevant elements, place, and time period. It is not defined merely by the file currently available.

---

### Statistical Inference

Statistical inference uses sample evidence to learn about a population.

* A parameter describes a population.

* A statistic is calculated from a sample.

* Estimation uses a sample statistic to estimate a population parameter.

* Hypothesis testing evaluates a claim about a population parameter.

Inference requires more than a large number of observations. The sample must connect credibly to the population, and uncertainty must be assessed.

---

```python
sample_a = cafe.sample(n=5, random_state=2301)
sample_b = cafe.sample(n=5, random_state=2302)

comparison = pd.DataFrame({
    "data": ["All 12 rows", "Sample A", "Sample B"],
    "observations": [len(cafe), len(sample_a), len(sample_b)],
    "mean_wait_minutes": [cafe["wait_minutes"].mean(), sample_a["wait_minutes"].mean(), sample_b["wait_minutes"].mean()],
    "mobile_share": [cafe["channel"].eq("Mobile").mean(), sample_a["channel"].eq("Mobile").mean(), sample_b["channel"].eq("Mobile").mean()]
})

display(comparison.round(3))
display(cafe.groupby("channel")["wait_minutes"].agg(["count", "mean"]).round(2))
```

### Analytics

Analytics transforms data into insight for decision making.

* Descriptive analytics: What happened?

* Predictive analytics: What is likely to happen, or how is one variable associated with another?

* Prescriptive analytics: Which action should be taken under stated objectives and constraints?

For café operations:

* Summarize waiting times by period.

* Predict congestion from order mix and staffing.

* Choose a staffing plan subject to cost and service targets.

The categories are connected, but a descriptive pattern alone does not determine the best action.

### Big Data, Data Mining, and Model Reliability

Big data are commonly described by:

* Volume: the amount of data

* Velocity: the speed of collection and processing

* Variety: the range of structured and unstructured data

Data mining uses statistical and computational methods to discover useful patterns and develop predictions from large data sets.

A model that fits the data used to build it may perform poorly on new data. 
-   Separating training data from test data helps evaluate whether a pattern is repeatable. 

-   Large data sets do not eliminate selection bias, measurement problems, overfitting, or misleading causal interpretations.
