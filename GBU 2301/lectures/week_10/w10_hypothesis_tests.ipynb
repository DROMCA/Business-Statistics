### Hypothesis Testing: Fill Volume

Alpha Bottling sets a target of 500 mL. Its quality team will pause the line for inspection if a sample provides sufficient evidence that the population mean fill is below target.

| Audit information | Value |
|---|---:|
| Target mean, $\mu_0$ | 500 mL |
| Known population standard deviation, $\sigma$ | 12 mL |
| Number of independently sampled bottles, $n$ | 36 |

* Assume the fill volumes come from a normal population.
* $\mu$ is the population mean for the current production process.
* Week 9 used a sample to estimate $\mu$. This audit uses a new sample to assess a specific claim about $\mu$.

---


### Null and Alternative Hypotheses

For the underfilling audit, the question is whether the population mean is below 500 mL.

$$H_0:\mu\geq500\text{ mL},\qquad
H_a:\mu<500\text{ mL}.$$

| Hypothesis | In this audit |
|---|---|
| Null hypothesis, $H_0$ | The population mean meets or exceeds the target. |
| Alternative hypothesis, $H_a$ | The population mean falls below the target. |

* The hypotheses are about the population parameter $\mu$.
* The sample mean $\bar x$ supplies the evidence.
* The equality $\mu=500$ belongs to $H_0$.

---


### One-Sided and Two-Sided Tests

Let $\mu_0$ be the proposed value of the population mean.

| Question about the population mean | Null hypothesis | Alternative hypothesis | Test direction |
|---|---|---|---|
| Is the mean below $\mu_0$? | $H_0:\mu\geq\mu_0$ | $H_a:\mu<\mu_0$ | Left-tailed |
| Is the mean above $\mu_0$? | $H_0:\mu\leq\mu_0$ | $H_a:\mu>\mu_0$ | Right-tailed |
| Does the mean differ from $\mu_0$? | $H_0:\mu=\mu_0$ | $H_a:\mu\ne\mu_0$ | Two-tailed |

The alternative determines which sample results count as evidence against $H_0$.

---


### Test Direction

Alpha Bottling could design different audits around the same 500 mL target.

| Purpose of the audit | Alternative | Sample results that support the alternative |
|---|---|---|
| Detect underfilling | $\mu<500$ | Sample means well below 500 mL |
| Detect costly overfilling | $\mu>500$ | Sample means well above 500 mL |
| Detect a shift in either direction | $\mu\ne500$ | Sample means far from 500 mL in either direction |

Choose the direction from the purpose of the audit before seeing the sample. Today's audit is designed to detect underfilling.

---


### The Null Boundary

For the underfilling audit, the null $H_0:\mu\geq500$ includes 500 mL and every higher mean.

* If the true mean is 500 mL, the sampling distribution of $\bar X$ is centered at 500.
* If the true mean is 502 mL, the distribution shifts to the right while its standard error stays the same.
* A sample mean at or below a fixed cutoff is therefore more likely at $\mu=500$ than at $\mu=502$.

$$P(\bar X\leq c\mid\mu=502)\leq
P(\bar X\leq c\mid\mu=500),\qquad\text{for a fixed cutoff }c.$$

The chance of a false alarm is largest at the null boundary, $\mu_0=500$. Use this value to calibrate the left-tailed test.

---


### Test Decisions

The quality team agrees on both actions before sampling.

| Test decision | Conclusion | Action |
|---|---|---|
| Reject $H_0$ | Sufficient evidence that $\mu<500$ mL | Pause the line for inspection |
| Fail to reject $H_0$ | Insufficient evidence that $\mu<500$ mL | Continue production and scheduled monitoring |

* The decision depends on how unusual the sample is under the null.
* Continuing production leaves uncertainty about the true population mean.

---


### Type I and Type II Errors

| Decision | $H_0$ is true: $\mu\geq500$ mL | $H_a$ is true: $\mu<500$ mL |
|---|---|---|
| Reject $H_0$ and pause | Type I error. Pause a process that meets the target. | Correct decision. Investigate underfilling. |
| Fail to reject $H_0$ and continue | Correct decision. Continue production. | Type II error. Underfilling continues. |

* Type I error can cost production time.
* Type II error can leave customers receiving underfilled bottles.

---


### Significance Level

The significance level $\alpha$ limits the probability of a Type I error. Choose it before sampling.

For Alpha Bottling, set $\alpha=0.05$. The test's rejection rule will be chosen so that

$$P(\text{reject }H_0\mid\mu=500)=0.05.$$

* Under this left-tailed rule, the chance of stopping the line is smaller when $\mu>500$.
* With repeated sampling at $\mu=500$, about 5% of these tests would incorrectly signal underfilling.
* A smaller $\alpha$ requires stronger sample evidence before pausing the line.

---


### Type II Error Probability

A Type II error occurs when the process underfills and the test fails to detect it. Specify a true mean below the target to describe its probability.

For example, if the true mean is 498 mL,

$$\beta(498)=P(\text{fail to reject }H_0\mid\mu=498).$$

* The probability can be different at another true mean, such as $\mu=495$ mL.
* For the same sample size and rejection rule, underfilling is easier to detect at 495 mL than at 498 mL.
* With sample size fixed, lowering $\alpha$ can make actual underfilling harder to detect.

---


### Sampling Distribution Under the Null

At the null boundary $\mu_0$,

$$E(\bar X)=\mu_0,\qquad \sigma_{\bar X}=\frac{\sigma}{\sqrt n}.$$

For Alpha Bottling, the null sampling distribution of $\bar X$ is normal with mean 500 mL and standard error

$$\sigma_{\bar X}=\frac{12}{\sqrt{36}}=2\text{ mL}.$$

The audit gives a sample mean of 496.4 mL. Standardize its distance from the null value:

$$z_{\text{obs}}=\frac{\bar x-\mu_0}{\sigma/\sqrt n}
=\frac{496.4-500}{12/\sqrt{36}}=-1.80.$$

This sample mean is 1.80 standard errors below the null mean.

---


### p-Value

The p-value is the probability, calculated under the null model, of a test statistic at least as extreme as the observed one in the direction specified by $H_a$.

For the left-tailed underfilling audit, smaller sample means give stronger evidence against $H_0$:

$$\text{p-value}=P(\bar X\leq496.4\mid\mu=500)
=P(Z\leq-1.80).$$

| Quantity | How it is obtained |
|---|---|
| Significance level $\alpha$ | Chosen before sampling to control the false-alarm rate |
| p-Value | Calculated from the observed sample using the null distribution |

Reject $H_0$ if the p-value is at most $\alpha$.

### Population Mean: $\sigma$ Known

At the null value $\mu_0$, the sample mean has standard error $\sigma/\sqrt n$. The test statistic measures the distance from $\mu_0$ in standard errors.

$$Z=\frac{\bar X-\mu_0}{\sigma/\sqrt n},
\qquad
z_{\text{obs}}=\frac{\bar x-\mu_0}{\sigma/\sqrt n}.$$

* Use a random sample with independent observations.
* At $\mu=\mu_0$, $Z$ follows the standard normal distribution under normal sampling. For an approximately normal sample mean, this reference distribution is approximate.
* The population standard deviation $\sigma$ is known.

For Alpha Bottling, $H_0:\mu\geq500$, $H_a:\mu<500$, and $z_{\text{obs}}=-1.80$.

---


### Left-Tailed Test: p-Value

In the underfilling audit, sample means at or below 496.4 mL are at least as extreme as the observed one.

$$\text{p-value}=P(Z\leq z_{\text{obs}})
=\Phi(z_{\text{obs}})
=\Phi(-1.80)\approx0.0359.$$

At the null boundary, about 3.59% of samples of this size would have a mean of 496.4 mL or less.

$$0.0359<\alpha=0.05.$$

* Reject $H_0$. The sample provides evidence that the population mean fill is below 500 mL.
* Under the audit rule, pause the line for inspection.

---


### Left-Tailed Test: Critical Value

Choose the cutoff so the rejection region has probability $\alpha$ under the null boundary.

$$P(Z\leq-z_\alpha)=\alpha,
\qquad
\text{reject }H_0\text{ when }z_{\text{obs}}\leq-z_\alpha.$$

For $\alpha=0.05$, the positive upper-tail critical value is $z_{0.05}=1.645$. The left-tail cutoff is $-1.645$.

$$z_{\text{obs}}=-1.80<-1.645.$$

The critical-value method also rejects $H_0$.

In fill-volume units, the cutoff for $\bar x$ is

$$c=\mu_0-z_\alpha\frac{\sigma}{\sqrt n}
=500-1.645(2)=496.71\text{ mL}.$$

The observed mean, 496.4 mL, is below this cutoff.

---


### Right-Tailed Test

Consider a separate audit designed to detect overfilling. It uses the same target, known standard deviation, and sample size. Suppose this audit gives $\bar x=503.6$ mL.

$$H_0:\mu\leq500,\qquad H_a:\mu>500.$$

$$z_{\text{obs}}=\frac{\bar x-\mu_0}{\sigma/\sqrt n}
=\frac{503.6-500}{12/\sqrt{36}}=1.80.$$

$$\text{p-value}=P(Z\geq z_{\text{obs}})
=1-\Phi(1.80)\approx0.0359.$$

At $\alpha=0.05$, both criteria give the same decision:

* $0.0359<0.05$.
* $z_{\text{obs}}=1.80>z_{0.05}=1.645$.

Reject $H_0$. The sample provides evidence of overfilling.

---


### Two-Tailed Test: p-Value

Suppose the audit had been designed in advance to detect a departure from 500 mL in either direction.

$$H_0:\mu=500,\qquad H_a:\mu\ne500.$$

With $\bar x=496.4$ mL, $z_{\text{obs}}=-1.80$. A departure of at least 1.80 standard errors can occur in either tail.

$$\begin{aligned}
\text{p-value}
&=P(Z\leq-|z_{\text{obs}}|)+P(Z\geq|z_{\text{obs}}|)\\
&=2[1-\Phi(|z_{\text{obs}}|)]
=2[1-\Phi(1.80)]\approx0.0719.
\end{aligned}$$

At $\alpha=0.05$, fail to reject $H_0$. For this audit, the sample gives insufficient evidence that the mean differs from 500 mL.

---


### Two-Tailed Test: Critical Values

Split $\alpha$ equally between the two tails.

$$\text{Reject }H_0\text{ when }z_{\text{obs}}\leq-z_{\alpha/2}
\quad\text{or}\quad z_{\text{obs}}\geq z_{\alpha/2}.$$

For $\alpha=0.05$, each tail has area 0.025. The cutoffs are $\pm1.96$, so $z_{\text{obs}}=-1.80$ lies between them.

| Purpose set before sampling | p-Value for $\bar x=496.4$ mL | Decision at 5% |
|---|---:|---|
| Detect a mean below 500 mL | 0.0359 | Reject $H_0$ |
| Detect a mean different from 500 mL | 0.0719 | Fail to reject $H_0$ |

In the two-tailed test, the lower rejection region has half the area of the left-tailed test's rejection region.

---


```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import Markdown, display
from scipy.stats import norm


def z_test_values(direction, z_obs, alpha):
    if direction == "Left-tailed":
        return norm.cdf(z_obs), (z_obs,), (-norm.isf(alpha),)
    if direction == "Right-tailed":
        return norm.sf(z_obs), (z_obs,), (norm.isf(alpha),)
    return 2 * norm.sf(abs(z_obs)), (-abs(z_obs), abs(z_obs)), (
        -norm.isf(alpha / 2), norm.isf(alpha / 2)
    )


def z_test_explorer():
    display(Markdown("### p-Value and Rejection Region"))
    style = {"description_width": "initial"}
    direction = widgets.Dropdown(
        options=["Left-tailed", "Right-tailed", "Two-tailed"],
        description="Test direction:", style=style
    )
    z_obs = widgets.FloatSlider(
        value=-1.80, min=-3.5, max=3.5, step=0.05, readout_format=".2f",
        description="Observed z:", continuous_update=False, style=style
    )
    alpha = widgets.Dropdown(
        options=[("1%", 0.01), ("5%", 0.05), ("10%", 0.10)], value=0.05,
        description="Significance level:", style=style
    )
    display(widgets.VBox([widgets.HBox([direction, alpha]), z_obs]))
    graph_handle = text_handle = None

    def shade_mask(x, cutoffs):
        if direction.value == "Left-tailed":
            return x <= cutoffs[0]
        if direction.value == "Right-tailed":
            return x >= cutoffs[0]
        return (x <= cutoffs[0]) | (x >= cutoffs[1])

    def update(change=None):
        nonlocal graph_handle, text_handle
        value, level = z_obs.value, alpha.value
        probability, observed_tails, critical_tails = z_test_values(
            direction.value, value, level
        )
        fig, axes = plt.subplots(1, 2, figsize=(11, 3.6), sharey=True,
                                 constrained_layout=True)
        for ax, title, cutoffs, color in [
            (axes[0], f"p-Value = {probability:.4f}", observed_tails, "#1f6f8b"),
            (axes[1], f"Rejection region area = {level:.2f}", critical_tails, "#d8872d"),
        ]:
            x = np.unique(np.r_[np.linspace(-4.1, 4.1, 1401), cutoffs])
            density = norm.pdf(x)
            ax.plot(x, density, color="#123b5d", linewidth=1.8)
            ax.fill_between(x, 0, density, where=shade_mask(x, cutoffs),
                            color=color, alpha=0.48)
            for cutoff in set(cutoffs):
                ax.axvline(cutoff, color=color, linewidth=1.1, linestyle=":")
            ax.axvline(value, color="#464646", linewidth=1.0, linestyle="--")
            ax.set(xlim=(-4.1, 4.1), ylim=(0, 0.43),
                   xlabel="Standard normal value", title=title)
            ax.grid(axis="y", alpha=0.15)
        axes[0].set_ylabel("Density")
        fig.suptitle(f"{direction.value} test   |   Dashed line: observed z = {value:.2f}")
        if graph_handle is None:
            graph_handle = display(fig, display_id=True)
        else:
            graph_handle.update(fig)
        plt.close(fig)

        if direction.value == "Left-tailed":
            expression = rf"P(Z\leq {value:.2f})=\Phi({value:.2f})"
        elif direction.value == "Right-tailed":
            expression = rf"P(Z\geq {value:.2f})=1-\Phi({value:.2f})"
        else:
            expression = rf"2P(Z\geq {abs(value):.2f})=2[1-\Phi({abs(value):.2f})]"
        cutoffs = ", ".join(f"{cutoff:.3f}" for cutoff in critical_tails)
        decision = "Reject" if probability <= level else "Fail to reject"
        content = Markdown(
            rf"$$\text{{p-value}}={expression}\approx {probability:.4f}.$$"
            f"\n\nCritical value(s): {cutoffs}. "
            rf"At $\alpha={level:.2f}$, {decision.lower()} $H_0$."
        )
        if text_handle is None:
            text_handle = display(content, display_id=True)
        else:
            text_handle.update(content)

    for control in (direction, z_obs, alpha):
        control.observe(update, names="value")
    update()
    return direction, z_obs, alpha


z_test_controls = z_test_explorer()
```

### Confidence Intervals and Two-Tailed Tests

For a two-tailed $z$ test at level $\alpha$, use the $100(1-\alpha)\%$ confidence interval with the same standard error.

$$\begin{aligned}
\mu_0\text{ lies strictly between the endpoints}
&\Longleftrightarrow
|\bar x-\mu_0|<z_{\alpha/2}\frac{\sigma}{\sqrt n}\\
&\Longleftrightarrow |z_{\text{obs}}|<z_{\alpha/2}.
\end{aligned}$$

At 95% confidence, the interval for the fill-volume audit is

$$\bar x\pm z_{\alpha/2}\frac{\sigma}{\sqrt n}
=496.4\pm1.96(2)=[492.48,\;500.32]\text{ mL}.$$

The null value, 500 mL, is inside the interval. The matching two-tailed test fails to reject $H_0$ at 5%.

### Tests for a Mean with Known $\sigma$

Use $z_{\text{obs}}=(\bar x-\mu_0)/(\sigma/\sqrt n)$ in each case.

| Alternative | p-Value | Rejection rule |
|---|---|---|
| $\mu<\mu_0$ | $\Phi(z_{\text{obs}})$ | $z_{\text{obs}}\leq-z_\alpha$ |
| $\mu>\mu_0$ | $1-\Phi(z_{\text{obs}})$ | $z_{\text{obs}}\geq z_\alpha$ |
| $\mu\ne\mu_0$ | $2[1-\Phi(\lvert z_{\text{obs}}\rvert)]$ | $\lvert z_{\text{obs}}\rvert\geq z_{\alpha/2}$ |

* The alternative sets the tail or tails.
* The significance level sets the rejection region.
* Use either the p-value or the critical-value rule and report the conclusion about the population mean.

### Population Mean: $\sigma$ Unknown

Estimate the standard error with the sample standard deviation $s$. This is the same estimated standard error used in the Week 9 $t$ interval.

$$s_{\bar x}=\frac{s}{\sqrt n}.$$

Under the null value $\mu_0$, use the statistic

$$T=\frac{\bar X-\mu_0}{S/\sqrt n},
\qquad
t_{\text{obs}}=\frac{\bar x-\mu_0}{s/\sqrt n}.$$

For independent observations from a normal population, the reference distribution is $t$ with $df=n-1$.

Use the tail specified by $H_a$. The $t$ distribution supplies the tail area and critical values.

---


### Conditions for a Mean Test

| Sampling situation | Implication for the test |
|---|---|
| Random, independent observations from a normal population | The $z$ test applies with known $\sigma$, and the $t$ test applies with unknown $\sigma$. |
| Large independent random sample with finite variance | The sampling distribution can be approximately normal. Strong skewness can require a larger sample. |
| Small sample with strongly skewed data or extreme outliers | The usual $t$ test can be unreliable. Review the data and the model. |
| Dependent observations or biased selection | The usual standard error or conclusion may be inappropriate. |

When $\sigma$ is unknown and the mean-test conditions hold, use $t$. As the sample size grows, its critical values approach those of the standard normal.

### One-Tailed $t$ Test: Service Time

Gamma Service investigates whether the average service time exceeds its 10-minute benchmark. Use the Week 9 sample of independent service times and assume an approximately normal population.

| Sample size | Sample mean | Sample SD | Significance level |
|---:|---:|---:|---:|
| 8 | 12 minutes | 2 minutes | 5% |

$$H_0:\mu\leq10,\qquad H_a:\mu>10.$$

$$t_{\text{obs}}=\frac{\bar x-\mu_0}{s/\sqrt n}
=\frac{12-10}{2/\sqrt8}\approx2.828,\qquad df=8-1=7.$$

$$\text{p-value}=P(T\geq2.828)\approx0.0127\qquad(df=7).$$

Reject $H_0$ at 5%. There is evidence that the average service time exceeds 10 minutes.

---


### One-Tailed $t$ Test: Critical Value

For the Gamma Service test, the rejection region is in the upper tail.

$$P(T\geq t_\alpha)=\alpha,\qquad df=n-1.$$

| Input to the $t$ table | Value |
|---|---:|
| Degrees of freedom | 7 |
| Upper-tail area | 0.05 |
| Critical value, $t_{0.05}$ | 1.895 |

$$t_{\text{obs}}\approx2.828>1.895.$$

Reject $H_0$. Excel gives the same critical value with `=T.INV.2T(0.10,7)`. The function's first argument is the total area in both tails, twice the upper-tail area used here.

---


### Two-Tailed $t$ Test: Service Time

For a test designed to detect a departure from 10 minutes in either direction,

$$H_0:\mu=10,\qquad H_a:\mu\ne10.$$

With the same sample, $t_{\text{obs}}\approx2.828$ and $df=7$.

$$\text{p-value}=2P(T\geq |t_{\text{obs}}|)
\approx2(0.01273)=0.0255.$$

At 5%, the cutoffs are $\pm t_{0.025}=\pm2.365$. Both the p-value and critical-value methods reject $H_0$.

The Week 9 interval was $[10.33,\;13.67]$ minutes at 95% confidence. It excludes the null value of 10 minutes, in agreement with the two-tailed test.

### p-Values in Excel

Enter the observed statistic in B5 and, for a $t$ test, the degrees of freedom in B6.

| Test direction | Standard normal statistic in B5 | $t$ statistic in B5 |
|---|---|---|
| Left-tailed | `=NORM.S.DIST(B5,TRUE)` | `=T.DIST(B5,B6,TRUE)` |
| Right-tailed | `=1-NORM.S.DIST(B5,TRUE)` | `=1-T.DIST(B5,B6,TRUE)` |
| Two-tailed | `=2*NORM.S.DIST(-ABS(B5),TRUE)` | `=T.DIST.2T(ABS(B5),B6)` |

`TRUE` requests cumulative probability. In the two-tailed row, the formula includes the probabilities of departures at least as large in both directions.

Use the unrounded test statistic to calculate the p-value. Compare the p-value with $\alpha$.

### Choosing Between $z$ and $t$

First determine how the standard deviation was obtained.

| Information available | Standard error used | Reference distribution |
|---|---|---|
| Known population SD $\sigma$ | $\sigma/\sqrt n$ | Standard normal |
| Sample SD $s$ from the same observations as $\bar x$ | $s/\sqrt n$ | $t$ with $n-1$ degrees of freedom |
| Earlier pilot SD used for sample-size planning | Use the current sample's $s$ to estimate the standard error | $t$ with $n-1$ degrees of freedom |

For the same nonzero statistic, the heavier $t$ tails give a larger p-value in a two-tailed test, or in a one-tailed test when the observed statistic points in the direction of $H_a$. The difference is most noticeable at small degrees of freedom.

### Population Proportion

Zeta Distribution monitors the share of orders dispatched on time. Its benchmark is 60%. The operations team wants to know whether the current process exceeds that benchmark.

| Quantity | In this audit |
|---|---|
| $p$ | Population proportion of orders dispatched on time |
| $p_0$ | Benchmark proportion, $0.60$ |
| $R$ | Number dispatched on time in the sample |
| $\hat p=R/n$ | Sample proportion dispatched on time |

$$H_0:p\leq0.60,\qquad H_a:p>0.60.$$

Evidence for $H_a$ comes from unusually high sample proportions. The same procedure uses the lower tail for $p<p_0$ and both tails for $p\ne p_0$.

---


### Standard Error Under the Null

For $n$ independent observations with success probability $p$, the count $R$ follows a binomial distribution and $\hat p=R/n$.

$$E(\hat p)=p,\qquad
\sigma_{\hat p}=\sqrt{\frac{p(1-p)}{n}}.$$

At the null boundary $p=p_0$, these become

$$E(\hat p)=p_0,\qquad
\sigma_{\hat p,0}=\sqrt{\frac{p_0(1-p_0)}{n}}.$$

If the null sampling distribution of $\hat p$ is approximately normal, the observed statistic is

$$z_{\text{obs}}=\frac{\hat p-p_0}{\sqrt{p_0(1-p_0)/n}}.$$

The test evaluates variation at the benchmark $p_0$. For the Week 9 confidence interval, we estimated the unknown standard error using $\hat p$.

---


### Conditions for a Proportion Test

* Use a random sample with independent observations, with the same success probability for each observation.
* Check the expected counts at the null value.

$$np_0\geq5,\qquad n(1-p_0)\geq5.$$

A quality team sets $\alpha=0.05$ to test whether the defect rate exceeds 2%. It finds eight defects among 100 independently sampled orders.

| Null expected count | Calculation | Meets the guideline? |
|---|---:|---|
| Defects | $100(0.02)=2$ | No |
| Orders without a defect | $100(0.98)=98$ | Yes |

Use the exact binomial distribution for this upper-tailed test. At $p_0=0.02$,

$$P(R\geq8)=1-P(R\leq7)\approx0.000932,\qquad R\sim\mathrm{Bin}(100,0.02).$$

Since $0.000932<0.05$, reject $H_0:p\leq0.02$. The sample provides evidence that the defect rate exceeds 2%.

### Right-Tailed Test: On-Time Dispatch

Zeta Distribution's sample consists of independent orders from its current process.

| Orders sampled | Dispatched on time | Benchmark | Significance level |
|---:|---:|---:|---:|
| 300 | 186 | 60% | 5% |

$$H_0:p\leq0.60,\qquad H_a:p>0.60.$$

$$np_0=300(0.60)=180,\qquad n(1-p_0)=300(0.40)=120.$$

$$\hat p=\frac{186}{300}=0.62,\qquad
z_{\text{obs}}=\frac{\hat p-p_0}{\sqrt{p_0(1-p_0)/n}}
=\frac{0.62-0.60}{\sqrt{0.60(0.40)/300}}\approx0.7071.$$

$$\text{p-value}=1-\Phi(z_{\text{obs}})\approx0.2398.$$

At 5%, fail to reject $H_0$. The sample gives insufficient evidence that the population on-time proportion exceeds 60%.

### Two-Tailed Test: Billing Errors

Kappa Billing has a historical error rate of 10%. It plans to investigate a change in either direction.

| Invoices independently sampled | Sampled invoices with an error | Benchmark error rate |
|---:|---:|---:|
| 400 | 24 | 10% |

$$H_0:p=0.10,\qquad H_a:p\ne0.10.$$

The null expected counts are $400(0.10)=40$ errors and $400(0.90)=360$ invoices without errors.

$$\hat p=\frac{24}{400}=0.06,\qquad
z_{\text{obs}}=\frac{0.06-0.10}{\sqrt{0.10(0.90)/400}}\approx-2.6667.$$

$$\text{p-value}=2[1-\Phi(|z_{\text{obs}}|)]\approx0.0077.$$

Reject $H_0$ at 5%. There is evidence that the population error rate differs from 10%. The observed change is downward, from the 10% benchmark to a sample estimate of 6%.

---


### Critical Values for a Proportion

For the right-tailed Zeta test, the 5% normal cutoff is $z_{0.05}\approx1.645$. The rejection rule is

$$z_{\text{obs}}\geq z_\alpha
\quad\Longleftrightarrow\quad
\hat p\geq p_0+z_\alpha\sqrt{\frac{p_0(1-p_0)}{n}}.$$

Substitute the test inputs:

$$\hat p\geq0.60+z_{0.05}\sqrt{\frac{0.60(0.40)}{300}}\approx0.6465.$$

With $\hat p=R/n$, multiply the unrounded proportion cutoff by 300. The count cutoff is about 193.96. Using the normal approximation, reject when at least 194 of the 300 orders are on time.

The sample had 186 on-time orders. The critical-value method gives the same decision as the p-value method.
